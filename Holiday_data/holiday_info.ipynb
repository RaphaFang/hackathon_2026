```python
import pandas as pd

# Load CSV 
df = pd.read_csv("Holidays_data.csv",sep=";",encoding="cp1252")

df.columns = df.columns.str.strip()
# Convert date
df["date"] = pd.to_datetime(df["date"], format="%d/%m/%Y")

df["day"] = df["date"].dt.day_name()

columns = list(df.columns)
columns.remove("day")
date_position = columns.index("date")
columns.insert(date_position + 1, "day")
df = df[columns]

# Show result
print(df)

# Save result
df.to_csv("Holidays_data_updated", index=False, encoding="utf-8-sig")
```
