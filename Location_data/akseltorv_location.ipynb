```python
import pandas as pd
import datetime as datetime

# Load the dataset
df = pd.read_csv("akseltorv_location.csv", sep=";")

# Clean column names
df.columns = df.columns.str.strip().str.lower()

# Clean text values
df = df.astype(str).apply(lambda col: col.str.strip())

# Print each place and opening hours
for index, row in df.iterrows():
    print("---------------")
    print(row["name"])
    print("Type:", row["type"])
    print("Distance:", row["distance_zone"])

    print("Monday:", row["open_mon"], "-", row["close_mon"])
    print("Tuesday:", row["open_tue"], "-", row["close_tue"])
    print("Wednesday:", row["open_wed"], "-", row["close_wed"])
    print("Thursday:", row["open_thu"], "-", row["close_thu"])
    print("Friday:", row["open_fri"], "-", row["close_fri"])
    print("Saturday:", row["open_sat"], "-", row["close_sat"])
    print("Sunday:", row["open_sun"], "-", row["close_sun"])

#save the cleaned dataset
df.to_csv('akseltorv_locations_cleaned_data.csv', index=False, encoding="utf-8-sig")

```

```python
# -------------------------------
# Question:
# Which places are open on Tuesday
# between 12:00 and 15:00
# in distance zones 0-50 and 50-100?
# -------------------------------

wanted_zones = ["0-50", "50-100"]
start_time = datetime.datetime.strptime("12:00", "%H:%M").time()
end_time = datetime.datetime.strptime("15:00", "%H:%M").time()

open_places = []

for index, row in df.iterrows():
    # Check distance zone
    if row["distance_zone"] not in wanted_zones:
        continue

    # Check if closed
    if row["open_tue"].lower() == "closed" or row["close_tue"].lower() == "closed":
        continue

    # Convert opening hours to time
    open_time = datetime.datetime.strptime(row["open_tue"], "%H:%M").time()
    close_time = datetime.datetime.strptime(row["close_tue"], "%H:%M").time()

    # Check if open for the whole period
    if open_time <= start_time and close_time >= end_time:
        open_places.append(row)

# Create result table
result = pd.DataFrame(open_places)

print("\nPlaces open on Tuesday from 12:00 to 15:00 in zones 0-50 and 50-100:")

if result.empty:
    print("No places found.")
else:
    print(result[["name", "type", "distance_zone", "open_tue", "close_tue"]])
```
